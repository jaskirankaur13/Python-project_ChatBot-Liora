```python
import tkinter as tk
from tkinter import scrolledtext
from openai import OpenAI

# 1. API Configuration
API_KEY = "YOUR_API_KEY"  

client = OpenAI(
    api_key=API_KEY,
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/"
)

# 2. Main Window Setup
root = tk.Tk()
root.title("Liora AI Chatbot")
root.geometry("400x500")

# 3. Chat Display Area
chat_area = scrolledtext.ScrolledText(root, wrap=tk.WORD, state='disabled')
chat_area.pack(padx=10, pady=10, fill=tk.BOTH, expand=True)

# 4. Function to Handle Sending Messages
def send_message():
    user_text = user_entry.get().strip()
    if not user_text:
        return

    # Show User Message
    chat_area.config(state='normal')
    chat_area.insert(tk.END, f"You: {user_text}\n\n")
    chat_area.config(state='disabled')
    user_entry.delete(0, tk.END)

    # Get Response from Gemini / OpenAI API
    try:
        response = client.chat.completions.create(
            model="gemini-3.6-flash",
            messages=[{"role": "user", "content": user_text}]
        )
        bot_reply = response.choices[0].message.content

        # Show Bot Message
        chat_area.config(state='normal')
        chat_area.insert(tk.END, f"Bot: {bot_reply}\n\n")
        chat_area.config(state='disabled')
        chat_area.yview(tk.END)
    except Exception as e:
        chat_area.config(state='normal')
        chat_area.insert(tk.END, f"Error: {str(e)}\n\n")
        chat_area.config(state='disabled')

# 5. User Input Text Field
user_entry = tk.Entry(root, font=("Arial", 12))
user_entry.pack(padx=10, pady=5, fill=tk.X)
user_entry.bind("<Return>", lambda event: send_message())

# 6. Send Button
send_btn = tk.Button(root, text="Send", command=send_message, bg="#4CAF50", fg="white")
send_btn.pack(padx=10, pady=5)

# 7. Run the App
root.mainloop()
```
